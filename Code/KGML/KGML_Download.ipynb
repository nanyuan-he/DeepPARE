```python
import os
import requests
```

```python
def download_kgml(save_dir="kgml_files",organism="hsa"):
    
    os.makedirs(save_dir, exist_ok=True)
    # 下载 KGML 文件
    url = f"https://rest.kegg.jp/list/pathway/{organism}"
    response = requests.get(url)
    if response.status_code != 200:
        raise ValueError(f"Failed to download KGML for {pathway_id}")
    

    pathway_list = response.text.strip().split("\n")
    for pathway in pathway_list:
        pathway_id = pathway.split("\t")[0].replace("path:", "")
        kgml_url = f"https://rest.kegg.jp/get/{pathway_id}/kgml"
        
        kgml_response = requests.get(kgml_url)
        if kgml_response.status_code == 200:
            file_path = os.path.join(save_dir,f"{pathway_id}.xml")
            with open(file_path, "w") as f:
                f.write(kgml_response.text)
            print(f"Downloaded: {pathway_id}")
        else:
            print(f"Failed to download KGML for {pathway_id}")
```
